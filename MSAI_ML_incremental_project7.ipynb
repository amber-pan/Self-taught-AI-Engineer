```python
# ============================================================
# CAPSTONE PROJECT
# Clustering Credit Card Users using PCA and K-Means
# Dataset: CC GENERAL.csv
# ============================================================

# ============================================================
# STEP 1 — Import Required Libraries
# ============================================================

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans

# Ignore warnings
import warnings
warnings.filterwarnings("ignore")


# ============================================================
# STEP 2 — Load Dataset
# ============================================================

# Update file path if needed
df = pd.read_csv("CC GENERAL.csv")

# Display first 5 rows
print(df.head())

# Dataset information
print("\nDataset Shape:", df.shape)

print("\nDataset Info:")
print(df.info())


# ============================================================
# STEP 3 — Check for Null Values
# ============================================================

print("\nNull Values:")
print(df.isnull().sum())

# ============================================================
# STEP 4 — Handle Missing Values
# ============================================================

# Remove customer ID column
df.drop("CUST_ID", axis=1, inplace=True)

# Fill missing values with column mean
df.fillna(df.mean(), inplace=True)

print("\nNull Values After Handling:")
print(df.isnull().sum())


# ============================================================
# STEP 5 — Feature Scaling using StandardScaler
# ============================================================

scaler = StandardScaler()

scaled_data = scaler.fit_transform(df)

# Convert back to DataFrame
scaled_df = pd.DataFrame(scaled_data, columns=df.columns)

print("\nScaled Data Sample:")
print(scaled_df.head())


# ============================================================
# STEP 6 — PCA with All Components
# ============================================================

pca_full = PCA()

pca_full.fit(scaled_df)

# Explained variance ratio
explained_variance = pca_full.explained_variance_ratio_

# Cumulative variance
cumulative_variance = np.cumsum(explained_variance)

print("\nExplained Variance Ratio:")
print(explained_variance)

print("\nCumulative Variance:")
print(cumulative_variance)


# ============================================================
# STEP 7 — Plot Number of Components vs Variance
# ============================================================

plt.figure(figsize=(10, 6))

plt.plot(
    range(1, len(cumulative_variance) + 1),
    cumulative_variance,
    marker='o'
)

plt.xlabel("Number of Principal Components")
plt.ylabel("Cumulative Explained Variance")
plt.title("PCA Explained Variance")
plt.grid(True)

plt.axhline(y=0.85, color='r', linestyle='--')

plt.show()


# ============================================================
# STEP 8 — Find Number of Components Covering 85% Variance
# ============================================================

n_components_85 = np.argmax(cumulative_variance >= 0.85) + 1

print("\nNumber of Components Covering 85% Variance:")
print(n_components_85)


# ============================================================
# STEP 9 — PCA with 2 Principal Components
# ============================================================

pca_2 = PCA(n_components=2)

pca_data = pca_2.fit_transform(scaled_df)

# Convert to DataFrame
pca_df = pd.DataFrame(
    data=pca_data,
    columns=["PCA1", "PCA2"]
)

print("\nPCA Transformed Data:")
print(pca_df.head())


# ============================================================
# STEP 10 — Visualize 2 Principal Components
# ============================================================

plt.figure(figsize=(10, 7))

sns.scatterplot(
    x="PCA1",
    y="PCA2",
    data=pca_df
)

plt.title("2D PCA Projection")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")

plt.show()


# ============================================================
# STEP 11 — Covariance Matrix from PCA
# ============================================================

covariance_matrix = pca_2.get_covariance()

cov_df = pd.DataFrame(
    covariance_matrix,
    columns=df.columns,
    index=df.columns
)

print("\nCovariance Matrix:")
print(cov_df)


# ============================================================
# STEP 12 — Find 2 Columns with Highest Covariance
# ============================================================

# Convert covariance matrix to absolute values
cov_abs = cov_df.abs()

# Remove diagonal values
np.fill_diagonal(cov_abs.values, 0)

# Find highest covariance pair
max_cov = cov_abs.unstack().sort_values(ascending=False)

highest_pair = max_cov.index[0]
highest_value = max_cov.iloc[0]

print("\nTwo Columns with Highest Covariance:")
print(highest_pair)

print("\nHighest Covariance Value:")
print(highest_value)


# ============================================================
# STEP 13 — Elbow Method for K-Means
# ============================================================

wcss = []

for i in range(1, 11):
    kmeans = KMeans(
        n_clusters=i,
        random_state=42
    )

    kmeans.fit(pca_df)

    wcss.append(kmeans.inertia_)

# Plot Elbow Graph
plt.figure(figsize=(10, 6))

plt.plot(range(1, 11), wcss, marker='o')

plt.title("Elbow Method")
plt.xlabel("Number of Clusters")
plt.ylabel("WCSS")

plt.grid(True)

plt.show()


# ============================================================
# STEP 14 — Apply K-Means with Ideal Clusters
# ============================================================

# Change this value after observing elbow graph
ideal_clusters = 4

kmeans_final = KMeans(
    n_clusters=ideal_clusters,
    random_state=42
)

clusters = kmeans_final.fit_predict(pca_df)

# Add cluster labels
pca_df["Cluster"] = clusters

print("\nCluster Counts:")
print(pca_df["Cluster"].value_counts())


# ============================================================
# STEP 15 — Visualize Clusters
# ============================================================

plt.figure(figsize=(12, 8))

sns.scatterplot(
    x="PCA1",
    y="PCA2",
    hue="Cluster",
    palette="Set1",
    data=pca_df,
    s=100
)

# Plot centroids
centroids = kmeans_final.cluster_centers_

plt.scatter(
    centroids[:, 0],
    centroids[:, 1],
    s=300,
    c='black',
    marker='X',
    label='Centroids'
)

plt.title("Customer Segments using PCA + KMeans")

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")

plt.legend()

plt.show()


# ============================================================
# STEP 16 — Interpretation of Results
# ============================================================

print("\nInterpretation Guide:")
print("""
1. PCA reduced the dimensionality of the dataset.
2. 2 principal components were used for visualization.
3. Customers were segmented into clusters using K-Means.
4. Similar customers are grouped together.
5. Businesses can use these groups for:
   - targeted marketing
   - customer segmentation
   - personalized offers
   - risk analysis
""")
```

```python
# ## Additional Step 16 — Compute and Plot Silhouette Scores

# ============================================================
#  Compute Silhouette Scores
# ============================================================

from sklearn.metrics import silhouette_score

silhouette_scores = []

K = range(2, 11)

for k in K:

    kmeans = KMeans(
        n_clusters=k,
        random_state=42
    )

    cluster_labels = kmeans.fit_predict(pca_df)

    score = silhouette_score(
        pca_df,
        cluster_labels
    )

    silhouette_scores.append(score)

    print(f"K = {k}, Silhouette Score = {score:.4f}")


# ============================================================
# Plot Silhouette Scores
# ============================================================

plt.figure(figsize=(10, 6))

plt.plot(
    K,
    silhouette_scores,
    marker='o'
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Score for Different K Values")

plt.grid(True)

plt.show()
```
